```python
import numpy as nump

n, p, m = 3, 4, 2

A = nump.random.rand(n, p)
B = nump.random.rand(p, m)

print("Matrix A (n x p):\n", A)
print("\nMatrix B (p x m):\n", B)

C = nump.dot(A, B)


print("\nDot product (A · B) → shape (n x m):\n", C)
```
